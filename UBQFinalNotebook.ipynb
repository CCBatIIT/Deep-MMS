```python
import mdtraj as md
import numpy as np

def create_arrow_bild_file(coordinates, output_file, color=(0, 0, 255), size=0.2, shape=1.0):
    """
    Create a .bild file with arrow representations for each atom.

    Parameters:
    - coordinates (numpy array): Atomic coordinates (n_atoms, 3).
    - output_file (str): The name of the output .bild file to be created.
    - color (tuple): RGB color values (default: (0, 0, 255) for blue).
    - size (float): Size of the arrows (default: 0.2).
    - shape (float): Shape of the arrows (default: 1.0).
    """
    with open(output_file, 'w') as bild_file:
        for coord in coordinates:
            r, g, b = color
            arrow_line = f".color {r} {g} {b}\n" \
                         f".arrow {coord[0] * 10} {coord[1] * 10} {coord[2] * 10} " \
                         f"{coord[0] * 10 + 10} {coord[1] * 10 + 10} {coord[2] * 10 + 10} {size} {shape}\n"
            bild_file.write(arrow_line)

# Example usage:
# Load your coordinates using mdtraj
pdb_file_path = '/Users/urvimekala/Downloads/1ubq.pdb'
traj = md.load(pdb_file_path)
coordinates = traj.xyz[0]  # Assuming you want the coordinates for the first frame

output_file_name = 'arrowscustomize-rgb.bild'
# Specify color as RGB values (0, 255, 0) for green
create_arrow_bild_file(coordinates, output_file_name, color=(0, 255, 0), size=0.3, shape=1.5)
```
