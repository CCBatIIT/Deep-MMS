```python
import mdtraj as md
import numpy as np
import os

def create_arrow_bild_file(start_coords, end_coords, output_file, color="blue", size=0.1, shape=0.3):
    """
    Create a .bild file with arrow representations for each atom showing movement from one trajectory to another.

    Parameters:
    - start_coords (numpy array): Atomic coordinates from the first trajectory (n_atoms, 3).
    - end_coords (numpy array): Atomic coordinates from the second trajectory (n_atoms, 3).
    - output_file (str): The name of the output .bild file to be created.
    - color (str): Color of the arrows (default: "blue").
    - size (float): Size of the arrows (default: 0.1).
    - shape (float): Shape of the arrows (default: 0.3).
    """
    with open(output_file, 'w') as bild_file:
        for start, end in zip(start_coords, end_coords):
            arrow_line = f".color {color}\n" \
                         f".arrow {start[0] * 10} {start[1] * 10} {start[2] * 10} " \
                         f"{end[0] * 10} {end[1] * 10} {end[2] * 10} {size} {shape}\n"
            bild_file.write(arrow_line)

# Load the reference pdb file
pdb_file_path = 'Simulation/ala_deca_peptide.pdb'
reference_traj = md.load(pdb_file_path)

###### IDENTIFIERS #####
latent_perturbed = 3    # Perturbed Latent working with
frame = 1-1             # Specify frame (remember to -1 since array starts at 0 but frame starts at 1)
color="yellow"             # Arrows color
# Load the two dcd trajectories
first_dcd_path = 'modelL/04_latents/rpt_2/ORIGINAL_modelL04.dcd'
second_dcd_path = 'modelL/04_latents/rpt_2/PERTURBED_LATENT_'+str(latent_perturbed)+'_added_0.5_modelL04.dcd'

# Reference dcd files on top of pdb files
first_traj = md.load(first_dcd_path, top=reference_traj.topology)
second_traj = md.load(second_dcd_path, top=reference_traj.topology)

# Latent 0: min is 933, max is 1407
# Latent 1: min is 1419, max is 1499
# Latent 2: min is 1786, max is 1496
# Latent 3: min is 880, max is 267
start_coordinates = first_traj.xyz[frame] 
end_coordinates = second_traj.xyz[frame]

# Make sure the output directory exists
output_directory = 'modelL/arrows'
os.makedirs(output_directory, exist_ok=True)

# Define the output file name with the directory
output_file_name = os.path.join(output_directory, 'Arrows_difference_latent_'+str(latent_perturbed)+'_frame_'+str(frame + 1)+'_'+color+'.bild')

# Call the function with the updated output file name
create_arrow_bild_file(start_coordinates, end_coordinates, output_file_name, color, size=0.1, shape=0.3)
```
